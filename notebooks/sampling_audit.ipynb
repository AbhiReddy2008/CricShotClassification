```python
import cv2
import numpy as np
import pandas as pd
import os
```

```python
def actual_frames_number(sample_frames_path):
    l = []
    actual_sample_frames = []
    output_folder = []
    for class_name in os.listdir(sample_frames_path):
        class_path = os.path.join(sample_frames_path,class_name)
        for video in os.listdir(class_path):
            video_frames_path = os.path.join(class_path, video)
            output_folder.append(f"{os.path.join('CricShotClassification','Dataset','processedData','frames',class_name,video)}")
            count = 0
            for sample_frames in os.listdir(video_frames_path):
                count += 1
            actual_sample_frames.append(count)
    l.append(actual_sample_frames)
    l.append(output_folder)
    return l


                
```

```python
def sample_frames_audit():
    rawData_excel = pd.read_csv(r"F:\CricShotClassification\Dataset\datasetAnalysis\raw_dataset_statistics.csv")
    data = []
    data.append(rawData_excel['class_name'].tolist())
    data.append(rawData_excel['video_name'].tolist())
    data.append(rawData_excel['frame_count'].tolist())
    sampling_target = [30 if i>=30 else i for i in rawData_excel['frame_count'].tolist()]
    sampling_method = ['uniform' if i>=30 else 'all' for i in rawData_excel['frame_count'].tolist()]
    data.append(sampling_target)
    data.append(sampling_method)
    data.append(sampling_target)
    actual_sampled_frames, output_folder = actual_frames_number('F:/CricShotClassification/Dataset/processedData/frames')
    data.append(actual_sampled_frames)
    data.append(list(np.array(data[5])-np.array(data[6])))
    data.append([True if i == 0 else False for i in data[len(data)-1]])
    data.append(rawData_excel['width'].tolist())
    data.append(rawData_excel['height'].tolist())
    data.append(output_folder)
    return data
```

```python
data = sample_frames_audit()
df = pd.DataFrame({
    'class_name': data[0],
    'video_name' : data[1],
    'original_frame_count': data[2],
    'sampling_target': data[3],
    'sampling_method': data[4],
    'expected_sampled_frames': data[5],
    'actual_sampled_frames': data[6],
    'missing_frames': data[7],
    'frame_count_match': data[8],
    'frame_width': data[9],
    'frame_height': data[10],
    'output_folder': data[11]
    })

df.to_csv(r"F:\CricShotClassification\Dataset\datasetAnalysis/sampled_frames_audit.csv",index = False)
```
