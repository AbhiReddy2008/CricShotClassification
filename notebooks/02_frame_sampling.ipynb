```python
import numpy as np
import cv2 
import os
```

```python
def get_uniform_sample_indices(frame_count, target_count = 30):
    if frame_count >= target_count:
        indices = np.linspace(0, frame_count-1, target_count, dtype = int)
    else:
        indices = np.arange(frame_count)
    return indices
```

```python
def get_sample_frames(rawData_path):
    for class_name in os.listdir(rawData_path):
        class_path = os.path.join(rawData_path, class_name)
        for video in os.listdir(class_path):
            video_path = os.path.join(class_path,video)
            cap = cv2.VideoCapture(video_path)
            frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            frame_index = 0
            count = 1
            indices = get_uniform_sample_indices(frame_count)
            while True:
                ret , frame = cap.read()
                if ret:
                    if frame_index in indices:
                        filename = os.path.splitext(video)[0]
                        dir_path = f"F:/CricShotClassification/Dataset/processedData/frames/{class_name}/{filename}"
                        file_path = os.path.join(dir_path, f"{filename}_0{count}.png")
                        if not os.path.exists(dir_path):
                            os.makedirs(dir_path, exist_ok = True)
                        cv2.imwrite(file_path, frame)
                        count += 1
                else :
                    break
                frame_index += 1
            cap.release()
                    
```

```python
get_sample_frames(f"F:/CricShotClassification/Dataset/rawData")
```
